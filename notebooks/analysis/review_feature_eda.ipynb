```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 및 마이너스 기호 깨짐 방지 설정
plt.rc('font', family='Malgun Gothic') # Mac 사용자라면 'AppleGothic'으로 변경
plt.rcParams['axes.unicode_minus'] = False
```

```python
df = pd.read_csv('../data/KoELECTRA_X_feature_v2.3_weighted.csv')
df.head()
```

```python
df.columns
```

```python
base_columns = ['매장명', '카테고리', '블루리본 여부']
weighted_columns = [col for col in df.columns if col.endswith('_weighted_score')]
selected_columns = base_columns + weighted_columns
selected_columns
```

```python
new_df = df[selected_columns]
new_df.head()
```

```python
##### score 분포 확인

new_df = df.iloc[:, :16].head()

# 1. _score로 끝나는 컬럼들만 리스트로 추출
weighted_cols = [col for col in new_df.columns if col.endswith('_score')]

# 2. Seaborn 시각화를 위해 데이터를 길게 변환 (Melt)
# 매장명, 카테고리, 블루리본여부는 그대로 두고 가중치 점수들만 아래로 길게 나열합니다.
df_melted = new_df.melt(id_vars=['매장명', '카테고리', '블루리본 여부'], 
                        value_vars=weighted_cols, 
                        var_name='평가지수', 
                        value_name='가중점수')

# 지수 이름에서 '_score'를 지워주면 차트가 더 깔끔해집니다.
df_melted['평가지수'] = df_melted['평가지수'].str.replace('_score', '')

# 3. Boxplot 그리기
plt.figure(figsize=(14, 6))
sns.boxplot(data=df_melted, x='평가지수', y='가중점수', palette='Set3')
plt.title('각 지수별 감성분석 긍정 점수 분포', fontsize=16)
plt.xticks(rotation=45) # X축 글씨가 겹치지 않게 45도 회전
plt.tight_layout()
plt.show()
```
