```python
# 데이터 경로는 본인 로컬에 맞게 설정
blue_review = pd.read_csv("../data/blueribbon_naver_reviews.csv")
blue_review.head()
```

### 한식 EDA

```python
k_review = blue_review.loc[blue_review.loc[:, '매장명'].isin(k_df.loc[:, '매장명'])]
print(len(k_review))
print(len(k_review.loc[:, '매장명'].unique()))
k_review.head()
```

```python
# 모든 리뷰 내용 하나의 긴 텍스트로 합치기
all_reviews = " ".join(k_review["리뷰 내용"].dropna().tolist())
```

```python
# 3. 형태소 분석을 통해 명사 추출
okt = Okt()
nouns = okt.nouns(all_reviews)
nouns[:10]  # 추출된 명사 중 일부 확인
```

```python
# 4. 불용어(Stopwords) 제거 및 한 글자 단어 제외
stop_words = ['정말', '진짜', '너무', '매우', '가장', '여기', '저기', '이것', '오늘']
filtered_nouns = [word for word in nouns if len(word) > 1 and word not in stop_words]
```

```python
# 5. 단어 빈도수 계산
word_counts = Counter(filtered_nouns)
print(f"[한식당] 상위 10개 키워드:", word_counts.most_common(10))
```

```python
wc = WordCloud(
    font_path=font_path,
    width=800, 
    height=400, 
    background_color='white',
    colormap='viridis'
).generate_from_frequencies(word_counts)
```

```python
# 7. 시각화
plt.figure(figsize=(10, 5))
plt.imshow(wc, interpolation='bilinear')
plt.axis('off')  # 축 숨기기
plt.title(f"[한식] 리뷰 워드클라우드", fontsize=15)
plt.show()
```
