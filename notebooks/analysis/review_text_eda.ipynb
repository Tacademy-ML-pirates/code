# 파일 정보

- 설명: 전체 리뷰 텍스트의 빈도, 키워드, 워드클라우드와 블루리본 집단 차이를 탐색합니다.
- 작성자: 김동혁

```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from konlpy.tag import Okt
from collections import Counter
from tqdm import tqdm
```

# 전체

```python
all_review = pd.read_excel('../data/merge_blue.xlsx', index_col=0)
all_review.head()
```

```python
print(len(all_review))
print(len(all_review.loc[:, '매장명'].unique()))
```

```python
review = all_review[all_review['리뷰 내용'].notna()].copy()
```

```python
drop_index = review.loc[review.loc[:, '리뷰 내용']=='텍스트 리뷰 없음'].index
print(len(drop_index))
```

```python
# 삭제할 행의 인덱스를 찾아서 drop에 전달
review = review.drop(drop_index)
```

```python
print(len(review))
```

```python
# # 모든 리뷰 내용 하나의 긴 텍스트로 합치기
# all_reviews = " ".join(review["리뷰 내용"].dropna().tolist())
```

```python
# 3. 형태소 분석을 통해 명사 추출
okt = Okt()
all_nouns = [] # 추출된 모든 명사를 담을 빈 리스트 생성

# 리뷰 텍스트를 하나씩 가져와서 형태소 분석 수행
for text in tqdm(review["리뷰 내용"].dropna()):
    nouns = okt.nouns(text)
    all_nouns.extend(nouns) # 개별 리뷰에서 추출된 명사들을 전체 리스트에 추가

print(all_nouns[:10])  # 추출된 명사 중 일부 확인
```

```python
# 4. 불용어(Stopwords) 제거 및 한 글자 단어 제외
# stop_words = ['정말', '진짜', '너무', '매우', '가장', '여기', '저기', '이것', '오늘']
# filtered_nouns = [word for word in nouns if len(word) > 1 and word not in stop_words]
filtered_nouns = all_nouns
```

```python
# 5. 단어 빈도수 계산
word_counts = Counter(filtered_nouns)
print(f"[리뷰] 상위 10개 키워드:", word_counts.most_common(10))
```

```python
plt.rc('font', family='Malgun Gothic') # 윈도우 사용자 (맑은 고딕)
plt.rcParams['axes.unicode_minus'] = False # 마이너스 기호 깨짐 방지
```

```python
# 6. 워드클라우드 생성 (※ 주의: 한글 폰트 경로를 자신의 PC에 맞게 설정해야 합니다)
# 윈도우 폰트 예시: 'C:/Windows/Fonts/malgun.ttf' (맑은 고딕)
# 맥 폰트 예시: '/System/Library/Fonts/AppleGothic.ttf' 또는 'AppleGothic'
font_path = 'C:/Windows/Fonts/malgun.ttf'  # 사용 중인 OS에 맞게 변경하세요.

wc = WordCloud(
    font_path=font_path,
    width=800, 
    height=400, 
    background_color='white',
    colormap='viridis'
).generate_from_frequencies(word_counts)
```

```python
# 7. 시각화
plt.figure(figsize=(10, 5))
plt.imshow(wc, interpolation='bilinear')
plt.axis('off')  # 축 숨기기
plt.title(f"[리뷰] 워드클라우드", fontsize=15)
plt.show()
```
