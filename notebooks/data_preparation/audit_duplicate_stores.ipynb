```python
# 1. 중복된 행이 총 몇 개인지 개수 확인
duplicate_count = df.duplicated().sum()
print(f"완전히 동일한 중복 행의 개수: {duplicate_count}개\n")
```

```python
# 2. 중복된 행들의 실제 데이터 추출
# keep=False 옵션을 주면 원본과 중복된 복사본을 모두 보여줍니다.
duplicates = df[df.duplicated(keep=False)]
```

```python
# 3. 중복 데이터 출력 (정렬해서 보면 똑같은 행끼리 묶여서 보기 편합니다)
# 매장명과 유저ID를 기준으로 정렬하여 출력합니다.
if duplicate_count > 0:
    duplicates_sorted = duplicates.sort_values(by=['매장명', '유저ID'])
else:
    print("중복된 데이터가 없습니다.")
duplicates_sorted
```

```python
# 중복된 행을 제거하고 첫 번째 행만 남기기 (원본 데이터프레임에 바로 적용)
df.drop_duplicates(inplace=True)
```

### cp949 -> utf-8

```python
import pandas as pd

# 1. 파일 이름 설정 (본인의 실제 파일명으로 수정해주세요)
input_file = "no_blueribbon_한식_리뷰_추천순_v2.3.csv"   # 기존 cp949 파일
output_file = "no_blueribbon_한식_리뷰_추천순_utf8.csv" # 새로 저장될 utf-8 파일

try:
    # 2. cp949 형식으로 기존 파일 읽어오기
    # encoding_errors='ignore' 옵션은 만약 cp949로 읽을 수 없는 깨진 글자가 있어도
    # 에러를 내며 멈추지 않고 해당 글자만 무시한 채 계속 읽어오도록 합니다.
    df = pd.read_csv(input_file, encoding="cp949", encoding_errors="ignore")
    
    # 3. utf-8-sig 형식으로 새 파일에 저장하기 (엑셀 호환)
    df.to_csv(output_file, index=False, encoding="utf-8-sig")
    
    print(f"🎉 성공적으로 변환되었습니다! '{output_file}' 파일을 확인해 주세요.")

except FileNotFoundError:
    print("❌ 지정한 원본 파일을 찾을 수 없습니다. 파일명과 경로를 다시 확인해 주세요.")
except Exception as e:
    print(f"❌ 변환 중 오류가 발생했습니다: {e}")
```
